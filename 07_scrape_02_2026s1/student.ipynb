```python
from bs4 import BeautifulSoup
from bs4.element import Tag
from IPython.core.display import HTML
```

```python
with open('Holidays_2023.html', encoding='utf8') as f:
    html = f.read()
```

```python
soup = BeautifulSoup(html, "lxml")
print(soup.prettify()[:200])
```

```python
country_table = soup.find('table', {'class': 'country-table'})
HTML(str(country_table))
```

```python
rows = country_table.find_all('tr')
for row in rows:
    print(list(row.children))
```

```python
idxs_by_day = {
    'Monday': 0,
    'Tuesday': 1,
    'Wednesday': 2,
    'Thursday': 3,
    'Friday': 4,
    'Saturday': 5,
    'Sunday': 6,
}

res = [0 for _ in range(7)]

for row in rows[1:]:
    ls = list(row.children)
    
    type = ls[7].get_text()
    if type != 'National Holiday': continue

    day = ls[1].get_text()
    res[idxs_by_day[day]] += 1

res
```

```python
for row in rows[1:]:
    ls = list(row.children)
    
    name = ls[5].get_text()

    if 'Visakha Bucha' in name:
        day = ls[1].get_text()
        print(day)
        break
else:
    print('not found')
```
